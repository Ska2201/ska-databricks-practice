REST API - request will Location Code

response will be Address (String)

 

Output file should be Summary of Total Employees at a particular location and their total salary

 

2,10,  15080 CapitalOne Dr Henrico  VA, 22000

```python
import pandas as pd
import requests

data = [('Kiran','Krish',123,10,10000),
('Naresh','Reddy',124,10,12000),
('Jubin','Kothari',125,12,12000),
('Aashna','Patel',126,11,15000)]

df = pd.DataFrame(data, columns = ['FN','LN','EmpID','LocationId','Sal'])
def address_fn(row):
    # Example: Call REST API to get address for LocationId
    location_id = row['LocationId']
    response = requests.get(f"https://api.example.com/location/{location_id}")
    return response.json().get("address", "")

df['address'] = df.apply(address_fn, axis = 1)

summary_df = df.groupby('LocationId').agg(
    Total_Employees=('EmpID', 'count'),
    Total_Salary=('Sal', 'sum')
).reset_index()

summary_df.diaplay()
```
