```sql
SELECT * FROM ska_catalog.bronze.com_worker
```

```sql
WITH ranked_tbl AS (
  SELECT salary,
    DENSE_RANK() OVER (ORDER BY salary DESC) AS `rank`
  FROM ska_catalog.bronze.com_worker
)
SELECT salary FROM ranked_tbl
WHERE rank = 5
LIMIT 1
```

```sql
%python
import pandas as pd

df = spark.table("ska_catalog.bronze.com_worker").toPandas()
df['rank'] = df['salary'].rank(method='dense', ascending=False)
result_df = df[df['rank'] == 5][['salary']]
display(result_df)
```
