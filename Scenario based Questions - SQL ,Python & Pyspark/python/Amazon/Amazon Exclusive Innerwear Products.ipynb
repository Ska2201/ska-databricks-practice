# Amazon Exclusive Innerwear Products

Find products which are exclusive to only Amazon and therefore not sold at Top Shop and Macy's. Your output should include the product name, brand name, price, and rating.

Two products are considered equal if they have the same product name and same maximum retail price (mrp column).

```sql
%skip
%sql
CREATE TABLE ska_catalog.bronze.innerwear_amazon_com (product_name VARCHAR(255),mrp VARCHAR(50),price VARCHAR(50),pdp_url VARCHAR(255),brand_name VARCHAR(100),product_category VARCHAR(100),retailer VARCHAR(100),description VARCHAR(255),rating FLOAT,review_count INT,style_attributes VARCHAR(255),total_sizes VARCHAR(50),available_size VARCHAR(50),color VARCHAR(50));

CREATE TABLE ska_catalog.bronze.innerwear_macys_com (product_name VARCHAR(255),mrp VARCHAR(50),price VARCHAR(50),pdp_url VARCHAR(255),brand_name VARCHAR(100),product_category VARCHAR(100),retailer VARCHAR(100),description VARCHAR(255),rating FLOAT,review_count FLOAT,style_attributes VARCHAR(255),total_sizes VARCHAR(50),available_size VARCHAR(50),color VARCHAR(50));

CREATE TABLE ska_catalog.bronze.innerwear_topshop_com (product_name VARCHAR(255),mrp VARCHAR(50),price VARCHAR(50),pdp_url VARCHAR(255),brand_name VARCHAR(100),product_category VARCHAR(100),retailer VARCHAR(100),description VARCHAR(255),rating FLOAT,review_count FLOAT,style_attributes VARCHAR(255),total_sizes VARCHAR(50),available_size VARCHAR(50),color VARCHAR(50));
```

```sql
-- Find products which are exclusive to only Amazon and therefore not sold at Top Shop and Macy's. Your output should include the product name, brand name, price, and rating.
-- Two products are considered equal if they have the same product name and same maximum retail price (mrp column).

SELECT
  a.product_name,
  a.brand_name,
  a.price,
  ROUND(a.rating,1) AS `rating`
FROM
  ska_catalog.bronze.innerwear_amazon_com a
LEFT JOIN
  ska_catalog.bronze.innerwear_topshop_com b
ON
  a.product_name = b.product_name AND a.mrp = b.mrp
LEFT JOIN
  ska_catalog.bronze.innerwear_macys_com c
ON
  c.product_name = a.product_name AND a.mrp = c.mrp
WHERE
  b.product_name IS NULL AND c.product_name IS NULL
ORDER BY a.product_name
```

```sql
%python
import pandas as pd

df_innerwear_topshop_com = spark.table(' ska_catalog.bronze.innerwear_topshop_com').toPandas()
df_innerwear_amazon_com = spark.table('ska_catalog.bronze.innerwear_amazon_com').toPandas()
df_innerwear_macys_com = spark.table('ska_catalog.bronze.innerwear_macys_com').toPandas()
exclusive_amazon = df_innerwear_amazon_com.merge(
    df_innerwear_topshop_com[['product_name', 'mrp']],
    on=['product_name', 'mrp'],
    how='left',
    indicator='topshop'
).merge(
    df_innerwear_macys_com[['product_name', 'mrp']],
    on=['product_name', 'mrp'],
    how='left',
    indicator='macys'
)
display(exclusive_amazon)

result = exclusive_amazon[
    (exclusive_amazon['topshop'] == 'left_only') & (exclusive_amazon['macys'] == 'left_only')
][['product_name', 'brand_name', 'price', 'rating']].copy()

result['rating'] = result['rating'].round(1)
result = result.sort_values('product_name')

display(result)
```
