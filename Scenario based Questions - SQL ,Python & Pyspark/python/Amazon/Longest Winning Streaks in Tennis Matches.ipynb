# Longest Winning Streaks in Tennis Matches

You are given a table of tennis players and their matches that they could either win (W) or lose (L). Find the longest streak of wins. A streak is a set of consecutive won matches of one player. The streak ends once a player loses their next match. Output the ID of the player or players and the length of the streak.

```python
%sql
WITH rankedMatches AS (
  SELECT 
    player_id,
    match_date,
    match_result,
    ROW_NUMBER() OVER (PARTITION BY player_id ORDER BY match_date) - ROW_NUMBER() OVER (PARTITION BY player_id , match_result ORDER BY match_date) AS grp
  FROM ska_catalog2.bronze.players_results
  WHERE match_result = 'W'
),
winStreaks As (
  SELECT 
    player_id,
    count(*) AS streak_length
  FROM rankedMatches
  GROUP BY player_id , grp
),
longestStreaks AS (
  SELECT
    player_id,
    MAX(streak_length) AS max_streak
  FROM winStreaks
  GROUP BY player_id
),
longestOverallStreaks AS (
  SELECT MAX( max_streak) AS longest_streak FROM longestStreaks
)
SELECT 
  ls.player_id,
  ls.max_streak AS longest_streak
FROM longestStreaks ls
JOIN longestOverallStreaks lo
ON ls.max_streak = lo.longest_streak
```

```python
import pandas as pd

df_players_results = spark.table('ska_catalog2.bronze.players_results').toPandas()

filtered_players = df_players_results[
  df_players_results['match_result'] == 'W'
  ].copy()

filtered_players['row_num_player'] = filtered_players\
  .groupby('player_id')['match_date']\
  .rank(method='first')\
  .astype(int)

filtered_players['row_num_player_result'] = filtered_players\
  .groupby(['player_id', 'match_result'])['match_date']\
  .rank(method='first')\
  .astype(int)

filtered_players['grp'] = filtered_players['row_num_player'] - filtered_players['row_num_player_result']

df_rankedMatches = filtered_players[['player_id', 'match_date', 'match_result', 'grp']]\
  .sort_values(['player_id', 'match_date'], ascending=[True, True])

display(df_rankedMatches)

df_winStreaks = df_rankedMatches.groupby(['player_id', 'grp'])\
  .agg(streak_length=('player_id', 'count')).reset_index()

df_winStreaks = df_winStreaks[
    ['player_id','streak_length']
  ]
display(df_winStreaks)

df_longestStreaks = df_winStreaks.groupby(['player_id'])\
  .agg(max_streak = ('streak_length', 'max')).reset_index()

df_longestOverallStreaks = df_longestStreaks['max_streak'].max()
df_longestOverallStreaks = pd.DataFrame({'longest_streak': [df_longestOverallStreaks]})

df_topStreakPlayers = df_longestStreaks.merge(df_longestOverallStreaks,left_on = 'max_streak',right_on = 'longest_streak', how = 'inner')

df_topStreakPlayers = df_topStreakPlayers[['player_id','longest_streak']]
display(df_topStreakPlayers)
```
