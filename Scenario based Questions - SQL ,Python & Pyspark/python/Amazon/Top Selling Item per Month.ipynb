# Top Selling Item per Month

Find the best selling item for each month (no need to separate months by year) where the biggest total invoice was paid. The best selling item is calculated using the formula (unitprice * quantity). Output the month, the description of the item along with the amount paid.

```python
%sql
SELECT * FROM ska_catalog.bronze.online_retail
```

```python
%sql
WITH MonthSales AS (
  SELECT
    month(invoicedate) AS month,
    description,
    SUM(unitprice * quantity) AS `SALES_PER_MONTH`
  FROM ska_catalog.bronze.online_retail
  GROUP BY month(invoicedate),description
),
ranked_dataset AS (
  SELECT
    month,
    description,
    SALES_PER_MONTH,
    RANK() OVER (PARTITION BY month ORDER BY SALES_PER_MONTH DESC) AS RANK
  FROM MonthSales
)
SELECT
  month,
  description,
  ROUND(SALES_PER_MONTH)
FROM ranked_dataset
WHERE rank = 1
ORDER BY month
```
