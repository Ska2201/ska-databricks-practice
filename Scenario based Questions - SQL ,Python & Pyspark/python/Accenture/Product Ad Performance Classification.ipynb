#Product Ad Performance Classification#

Following a recent advertising campaign, the marketing department wishes to classify its efforts based on the total number of units sold for each product.

You have been tasked with calculating the total number of units sold for each product and categorizing ad performance based on the following criteria for items sold:

Outstanding: 30+
Satisfactory: 20 - 29
Unsatisfactory: 10 - 19
Poor: 1 - 9

Your output should contain the product ID, total units sold in descending order, and its categorized ad performance.

🔍By solving this, you'll learn how to use when, case and group by. Give it a try and share the output! 👇

```python
%skip
%sql

CREATE TABLE  ska_catalog.bronze.marketing_campaign (user_id INT,created_at TIMESTAMP,product_id INT,quantity INT,price INT);

INSERT INTO ska_catalog.bronze.marketing_campaign (user_id, created_at, product_id, quantity, price) VALUES(1, '2020-01-01', 101, 25, 200),(2, '2020-01-01', 102, 5, 150),(3, '2020-01-02', 103, 15, 300),(4, '2020-01-03', 101, 10, 200),(5, '2020-01-04', 102, 22, 150),(6, '2020-01-05', 104, 8, 120),(7, '2020-01-06', 105, 18, 250),(8, '2020-01-07', 101, 30, 200),(9, '2020-01-08', 103, 20, 300),(10, '2020-01-09', 104, 9, 120);
```

```python
%sql
SELECT * FROM ska_catalog.bronze.marketing_campaign
```

```python
%sql
-- Outstanding: 30+ Satisfactory: 20 - 29 Unsatisfactory: 10 - 19 Poor: 1 - 9
SELECT
  product_id,
  SUM(quantity) As `Total Units Sold`,
  CASE
    WHEN SUM(quantity) >= 30 THEN 'Outstanding'
    WHEN SUM(quantity) >= 20 AND SUM(quantity) <=29 THEN 'Satisfactory'
    WHEN  SUM(quantity) >= 10 AND SUM(quantity) <=19 THEN 'Unsatisfactory'
    ELSE 'Poor'
  END AS `Performance_Category`
FROM ska_catalog.bronze.marketing_campaign
GROUP BY product_id
ORDER BY product_id, 'Total Units Sold' DESC;
```

```python
import pandas as pd

df_Performance_Category = spark.table("ska_catalog.bronze.marketing_campaign").toPandas()
df_result = df_Performance_Category.groupby('product_id', as_index=False)\
    .agg({'quantity': 'sum'})\
    .rename(columns = {'quantity': 'Total_units_sold' })

def categorize(units):
    if units >= 30:
        return 'outstanding'
    elif (units >= 20) & (units <=29):
        return 'Satisfactory'
    elif units >=10 & units <=19:
        return 'Unsatisfactory'
    else:
        return 'Poor'

df_result['performance_category'] = df_result['Total_units_sold'].apply(categorize)
df_result.display()
```
