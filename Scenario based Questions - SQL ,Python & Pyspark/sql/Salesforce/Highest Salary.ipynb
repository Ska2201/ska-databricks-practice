```sql
-- Find the highest salary among salaries that appears only once.
WITH salary_cnt As (
  SELECT  salary, COUNT(*) As `salary_count`
  FROM ska_catalog.bronze.sales_force_emp
  GROUP BY  salary
)
SELECT MAX(salary) AS `Highest_salary`
FROM salary_cnt
WHERE salary_count = 1
```
