```sql
SELECT
  DATE_FORMAT(customer_placed_order_datetime, 'EEEE') AS weekday,
  HOUR(customer_placed_order_datetime) AS order_hour,
  ROUND(
    AVG(order_total + tip_amount - discount_amount - refunded_amount), 2
  ) AS average_earnings
FROM ska_catalog.bronze.doordash_delivery
GROUP BY
  DATE_FORMAT(customer_placed_order_datetime, 'EEEE'),
  HOUR(customer_placed_order_datetime)
ORDER BY
  weekday,
  order_hour
;
```
