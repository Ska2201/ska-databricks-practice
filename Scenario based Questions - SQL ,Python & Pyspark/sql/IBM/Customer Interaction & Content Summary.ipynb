# Customer Interaction & Content Summary

Calculate the total number of interactions and the total number of contents created for each customer. Include all interaction types and content types in your calculations. Your output should include the customer's ID, the total number of interactions, and the total number of content items.

🌀By solving this, you'll learn how to use coalesce, outer join, subquery. Give it a try and share the output! 👇

```python

%sql
CREATE TABLE ska_catalog2.bronze.customer_interactions (customer_id BIGINT, interaction_date TIMESTAMP, interaction_id BIGINT, interaction_type VARCHAR(50));

INSERT INTO ska_catalog2.bronze.customer_interactions (customer_id, interaction_date, interaction_id, interaction_type) VALUES (1, '2023-01-15 10:30:00', 101, 'Click'), (1, '2023-01-16 11:00:00', 102, 'Purchase'), (2, '2023-01-17 14:45:00', 103, 'View'), (3, '2023-01-18 09:20:00', 104, 'Share'), (3, '2023-01-18 09:25:00', 105, 'Like'), (4, '2023-01-19 12:10:00', 106, 'Comment');

CREATE TABLE ska_catalog2.bronze.user_contents (content_id BIGINT, content_text VARCHAR(255), content_type VARCHAR(50), customer_id BIGINT);

INSERT INTO ska_catalog2.bronze.user_contents (content_id, content_text, content_type, customer_id) VALUES (201, 'Welcome Post', 'Blog', 1), (202, 'Product Review', 'Review', 2), (203, 'Event Photos', 'Photo', 3), (204, 'Tutorial Video', 'Video', 3), (205, 'Survey Response', 'Survey', 4);
```

```python
%sql
SELECT
  ci.customer_id,
  COALESCE(COUNT(DISTINCT ci.interaction_id),0) AS total_interaction,
  COALESCE(COUNT(DISTINCT uc.content_id),0) AS total_content_created
FROM (SELECT customer_id ,interaction_id  FROM ska_catalog2.bronze.customer_interactions)  ci
FULL OUTER JOIN 
(SELECT customer_id, content_id FROM ska_catalog2.bronze.user_contents) uc
ON ci.customer_id = uc.customer_id
GROUP BY ci.customer_id;
```
