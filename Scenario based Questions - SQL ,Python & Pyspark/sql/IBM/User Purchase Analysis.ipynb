```sql
WITH q1 AS (
  SELECT
      user_id,
      date,
      amount_spent,
      day_name,
      WEEKOFYEAR(date) AS week_number
    FROM ska_catalog.bronze.user_purchases
    WHERE day_name = 'Friday'
      AND month(date) IN (1,2,3)
)
SELECT week_number, ROUND(AVG(amount_spent),2) AS Avg_amt_users 
FROM q1
GROUP BY week_number
ORDER BY week_number ASC, Avg_amt_users DESC; 
```
