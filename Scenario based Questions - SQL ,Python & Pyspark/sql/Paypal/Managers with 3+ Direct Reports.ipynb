##Managers with 3+ Direct Reports

Find managers with at least 3 direct reporting employees. In situations where user is reporting to himself/herself, count that also.
Output first names of managers.

```sql
SELECT
  CONCAT(INITCAP(b.first_name),' ', INITCAP(b.last_name) ) AS `MANAGER_NAME`,
  COUNT(a.id) As `COUNT`
FROM ska_catalog.bronze.paypal_emp a
JOIN ska_catalog.bronze.paypal_emp b
ON a.manager_id = b.id
GROUP BY MANAGER_NAME
HAVING COUNT(a.id) >= 5
```
