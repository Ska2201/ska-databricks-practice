## Product Transaction Analysis

Find the number of transactions that occurred for each product. Output the product name along with the corresponding number of transactions and order records by the product id in ascending order. You can ignore products without transactions.

```sql
-- Find the number of transactions that occurred for each product. Output the product name along with the corresponding number of transactions and order records by the product id in ascending order. You can ignore products without transaction

SELECT i.product_id, i.product_name,COUNT(t.transaction_id) AS `No_of_trans`
FROM ska_catalog.bronze.excel_sql_inventory_data i
LEFT JOIN ska_catalog.bronze.excel_sql_transaction_data t
ON i.product_id = t.product_id
GROUP BY  i.product_id, i.product_name
HAVING COUNT(t.transaction_id) >= 1
ORDER BY i.product_id;
```
