## Viewer-to-Streamer Session Analysis

From users who had their first session as a viewer, how many streamer sessions have they had? Return the user id and number of sessions in descending order. In case there are users with the same number of sessions, order them by ascending user id.

🌀By solving this, you'll learn how to use Cte, Group by, Join. Give it a try and share the output! 👇

```sql
CREATE TABLE ska_catalog2.bronze.twitch_sessions (user_id BIGINT, session_start TIMESTAMP, session_end TIMESTAMP, session_id BIGINT PRIMARY KEY, session_type VARCHAR(20) CHECK (session_type IN ('viewer', 'streamer')));

INSERT INTO ska_catalog2.bronze.twitch_sessions (user_id, session_start, session_end, session_id, session_type) VALUES (101, '2024-02-01 10:00:00', '2024-02-01 11:00:00', 1, 'viewer'), (101, '2024-02-02 14:00:00', '2024-02-02 15:30:00', 2, 'streamer'), (102, '2024-02-01 09:30:00', '2024-02-01 10:30:00', 3, 'viewer'), (102, '2024-02-03 16:00:00', '2024-02-03 17:00:00', 4, 'streamer'), (102, '2024-02-05 18:00:00', '2024-02-05 19:30:00', 5, 'streamer'), (103, '2024-02-02 11:00:00', '2024-02-02 12:00:00', 6, 'viewer'), (104, '2024-02-01 08:30:00', '2024-02-01 09:00:00', 7, 'viewer'), (104, '2024-02-04 20:00:00', '2024-02-04 21:00:00', 8, 'streamer'), (104, '2024-02-06 22:00:00', '2024-02-06 23:00:00', 9, 'streamer'), (104, '2024-02-07 15:00:00', '2024-02-07 16:30:00', 10, 'streamer');
```

```sql
SELECT * FROM ska_catalog2.bronze.twitch_sessions;
```

```sql
WITH FirstSession AS (
  SELECT user_id,
    MIN(session_start) As first_session_time
  FROM ska_catalog2.bronze.twitch_sessions
  WHERE session_type = 'viewer'
  GROUP BY user_id
)
SELECT ts.user_id ,
  COUNT(*) AS `streamer_session_count`
FROM ska_catalog2.bronze.twitch_sessions ts
JOIN FirstSession fs
ON ts.user_id = fs.user_id
WHERE ts.session_type = 'streamer'
GROUP BY ts.user_id
ORDER BY streamer_session_count DESC , ts.user_id ASC;
```
