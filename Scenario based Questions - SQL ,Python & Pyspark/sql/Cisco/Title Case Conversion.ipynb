## Title Case Conversion

Convert the first letter of each word found in content_text to uppercase, while keeping the rest of the letters lowercase. Your output should include the original text in one column and the modified text in another column.


```sql
%skip
CREATE TABLE ska_catalog.bronze.user_content (content_id INT PRIMARY KEY,customer_id INT,content_type VARCHAR(50),content_text VARCHAR(255));

INSERT INTO ska_catalog.bronze.user_content (content_id, customer_id, content_type, content_text) VALUES(1, 2, 'comment', 'hello world! this is a TEST.'),(2, 8, 'comment', 'what a great day'),(3, 4, 'comment', 'WELCOME to the event.'),(4, 2, 'comment', 'e-commerce is booming.'),(5, 6, 'comment', 'Python is fun!!'),(6, 6, 'review', '123 numbers in text.'),(7, 10, 'review', 'special chars: @#$$%^&*()'),(8, 4, 'comment', 'multiple CAPITALS here.'),(9, 6, 'review', 'sentence. and ANOTHER sentence!'),(10, 2, 'post', 'goodBYE!');
```

```sql
SELECT content_text,initcap(content_text) FROM ska_catalog.bronze.user_content
```

```sql
SELECT
  content_id,
  content_text,
  concat_ws(' ', collect_list(upper(left(value,1)) || lower(substring(value,2))))
    AS title_case_text
FROM ska_catalog.bronze.user_content
LATERAL VIEW explode(split(content_text, ' ')) AS value
GROUP BY content_id, content_text
```

In Oracle, use **STRING_AGG** and **CROSS APPLY**:

sql

SELECT 

  CONTENT_ID,

  CONTENT_TEXT AS ORIGINAL_TEXT,

  STRING_AGG(
    UPPER(LEFT(value, 1)) + LOWER(SUBSTRING(value, 2, LEN(value))),
    ' '
  ) AS MODIFIED_TEXT

FROM ska_catalog.bronze.user_content

CROSS APPLY STRING_SPLIT(CONTENT_TEXT, ' ')

GROUP BY CONTENT_ID, CONTENT_TEXT;
