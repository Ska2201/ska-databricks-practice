```sql
WITH dates AS (
    SELECT
      user_id,
      date AS current_date,
      lag(date) OVER (PARTITION BY user_id ORDER BY date) AS previous_date,
      lead(date) OVER (PARTITION BY user_id ORDER BY date) AS next_date
    FROM ska_catalog.bronze.sf_events
)
SELECT DISTINCT
  user_id
FROM dates
WHERE 
  date_diff(current_date, previous_date) = 1 AND  
  date_diff(next_date, current_date) = 1
ORDER BY user_id;
```
