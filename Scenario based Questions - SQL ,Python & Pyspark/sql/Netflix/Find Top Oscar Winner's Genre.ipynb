𝐄𝐱𝐩𝐥𝐚𝐧𝐚𝐭𝐢𝐨𝐧 𝐭𝐨 𝐒𝐨𝐥𝐯𝐞 𝐐𝐮𝐞𝐫𝐲
1. WinnerCount CTE: Calculates the total Oscar wins for each nominee by counting rows where winner = 1.

2. Final Selection:
The TOP 1 clause fetches all rows with the highest total_wins, sorted alphabetically by name to handle ties.
We join the WinnerCount CTE with nominee_information on the nominee’s name to retrieve the top_genre for the top nominee(s) in terms of Oscar wins.

```sql
WITH winner_cnt AS (
  SELECT o.nominee,
    COUNT(*) AS total_wins
  FROM ska_catalog.bronze.oscar_nominees o
  WHERE winner = 1
  GROUP BY o.nominee
)
SELECT ni.top_genre
FROM winner_cnt wc
JOIN ska_catalog.bronze.nominee_information ni
ON ni.name = wc.nominee
ORDER BY ni.name ASC, wc.total_wins DESC
LIMIT 1 ;
```
