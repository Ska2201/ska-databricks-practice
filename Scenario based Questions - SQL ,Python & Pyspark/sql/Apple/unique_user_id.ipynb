```sql
-- Write a query that returns the number of unique users per client per month
SELECT
  client_id AS `CLIENT_ID`,
  date_format(time_id, 'yyyy-MM') AS `YEAR_MONTH`,
  collect_list(user_id) AS `USERS_LIST`, --Returns an array of values from a column for each group.
  collect_set(user_id) AS `UNIQUE_USERS_LIST`, --Returns an  array of unique values from a column for each group.
  COUNT(DISTINCT user_id ) AS `NO_OF_USERS`
FROM ska_catalog.bronze.fact_events
GROUP BY CLIENT_ID,date_format(time_id, 'yyyy-MM')
ORDER BY CLIENT_ID,YEAR_MONTH ;
```
