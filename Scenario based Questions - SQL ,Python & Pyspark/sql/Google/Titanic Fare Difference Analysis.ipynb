# Titanic Fare Difference Analysis


Find the average absolute fare difference between a specific passenger and all passengers that belong to the same pclass, both are non-survivors and age difference between two of them is 5 or less years. Do that for each passenger (that satisfy above mentioned coniditions). Output the result along with the passenger name.

```sql
SELECT * FROM ska_catalog2.bronze.titanic 
```

```sql
WITH passesngerFareDiff As (
  SELECT
    t1.name AS passesger_name,
    t1.passengerid AS passenger_id,
    ABS(t1.fare - t2.fare) AS fare_difference
  FROM ska_catalog2.bronze.titanic t1
  JOIN
    ska_catalog2.bronze.titanic t2
  ON
    t1.pclass = t2.pclass AND
    t1.survived = 0 AND t2.survived = 0 AND
    ABS (t1.age - t2.age)  <= 5 AND
    t1.passengerid != t2.passengerid
)
SELECT 
  t1.name AS `passenger_name`,
  AVG(pfd.fare_difference) AS `avg_abs_fare_diff`
FROM ska_catalog2.bronze.titanic t1
LEFT JOIN passesngerFareDiff pfd
ON t1.passengerid = pfd.passenger_id
WHERE t1.survived = 0
GROUP BY t1.name
ORDER BY  `passenger_name`
```
