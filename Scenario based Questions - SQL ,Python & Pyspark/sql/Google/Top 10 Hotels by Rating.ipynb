```sql
SELECT * FROM ska_catalog.bronze.hotel_address
```

```sql
-- Find the three ten hotels with the highest ratings. Output the hotel name along with the corresponding average score. Sort records based on the average score in descending order.
WITH ranked_hotel (
  SELECT hotel_name, average_score,
    DENSE_RANK() OVER (ORDER BY average_score DESC) AS `rank`
  FROM ska_catalog.bronze.hotel_address
)
SELECT hotel_name, ROUND(average_score,1) AS `AVG_SCORE` 
FROM ranked_hotel
WHERE rank <= 3
ORDER BY rank;
```
