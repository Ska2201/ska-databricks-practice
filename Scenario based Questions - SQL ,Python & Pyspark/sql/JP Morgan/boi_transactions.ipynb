Monday - Friday 09:00 - 16:00  
Saturday & Sunday Closed  
Irish Public Holidays 25th and 26th December

```sql
SELECT transaction_id
FROM ska_catalog.bronze.boi_transactions
WHERE MONTH(time_stamp) = 12
  AND YEAR(time_stamp) = 2022
  AND (
    -- Saturday (6) or Sunday (0)
    DAYOFWEEK(time_stamp) IN (1, 7)
    OR
    -- Public holidays
    (DAY(time_stamp) IN (25, 26))
    OR
    -- Weekday but outside business hours
    (
      DAYOFWEEK(time_stamp) BETWEEN 2 AND 6
      AND (
        HOUR(time_stamp) < 9
        OR HOUR(time_stamp) > 16
        OR (HOUR(time_stamp) = 16 AND MINUTE(time_stamp) > 0)
      )
    )
  )
```
