# Most Profitable Location Analysis


Find the most profitable location. Write a query that calculates the average signup duration and average transaction amount for each location, and then compare these two measures together by taking the ratio of the average transaction amount and average duration for each location.

Your output should include the location, average duration, average transaction amount, and ratio. Sort your results from highest ratio to lowest.

```sql
%skip
CREATE TABLE ska_catalog.bronze.cust_signups (location VARCHAR(100), plan_id BIGINT, signup_id BIGINT PRIMARY KEY, signup_start_date TIMESTAMP, signup_stop_date TIMESTAMP);

INSERT INTO ska_catalog.bronze.cust_signups (location, plan_id, signup_id, signup_start_date, signup_stop_date) VALUES  ('New York', 101, 1, '2025-01-01', '2025-01-31'), ('San Francisco', 102, 2, '2025-01-05', '2025-02-05'), ('Los Angeles', 103, 3, '2025-01-10', '2025-01-20'), ('New York', 104, 4, '2025-02-01', '2025-02-28'), ('Los Angeles', 105, 5, '2025-01-15', '2025-01-25');

CREATE TABLE ska_catalog.bronze.cust_transactions (amt FLOAT, signup_id BIGINT, transaction_id BIGINT PRIMARY KEY, transaction_start_date TIMESTAMP);

INSERT INTO ska_catalog.bronze.cust_transactions (amt, signup_id, transaction_id, transaction_start_date) VALUES (100.50, 1, 1001, '2025-01-10'), (200.75, 1, 1002, '2025-01-20'), (150.00, 2, 1003, '2025-01-15'), (300.00, 3, 1004, '2025-01-12'), (400.00, 4, 1005, '2025-02-15'), (250.00, 5, 1006, '2025-01-20');
```

```sql
SELECT
  s.location,
  AVG(DATEDIFF(DAY, s.signup_start_date, s.signup_stop_date)) AS avg_duration,
  AVG(t.amt) AS avg_transaction_amount,
  CASE
    WHEN AVG(DATEDIFF(DAY, s.signup_start_date,s.signup_stop_date))>0 THEN AVG(t.amt)/ AVG(DATEDIFF(DAY, s.signup_start_date,s.signup_stop_date)) 
  ELSE 0
  END AS ratio
FROM
  ska_catalog.bronze.cust_signups s
JOIN
  ska_catalog.bronze.cust_transactions t
ON s.signup_id = t.signup_id
GROUP BY s.location
ORDER BY ratio DESC;
```
